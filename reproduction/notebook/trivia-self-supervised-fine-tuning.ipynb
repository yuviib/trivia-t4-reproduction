````python
%%writefile patch_reward.py
import re

path = "/kaggle/working/TRivia/training/exps/trivia_reward_plugin.py"
with open(path, "r", encoding="utf-8") as f:
    content = f.read()

pattern = re.compile(
    r"    def __init__\(self, args=None, \*\*kwargs\) -> None:.*?\n(?=    def )",
    re.DOTALL,
)

match = pattern.search(content)
if not match:
    print("----- COULD NOT MATCH. FIRST 3000 CHARS OF FILE: -----")
    print(content[:3000])
    raise SystemExit("Could not locate __init__ block, see printed content above")

new_init = (
    "    def __init__(self, args=None, **kwargs) -> None:\n"
    "        super().__init__(args)\n"
    "        self.clients = [OpenAI(api_key=os.environ[\"OPENAI_API_KEY\"])]\n\n"
)

content = content[:match.start()] + new_init + content[match.end():]

with open(path, "w", encoding="utf-8") as f:
    f.write(content)
print("Patched trivia_reward_plugin.py successfully")
````

````python
# =====================================================================
# GRPO LoRA Before/After Comparison — Qwen2.5-VL-3B-Instruct (Table->HTML)
# Run this as a single Kaggle notebook cell. Requires OPENAI_API_KEY set.
# =====================================================================

import os
import sys
import types

# Must be set before `transformers` is imported anywhere below — this Kaggle
# image has a broken TensorFlow/protobuf install, and without USE_TF=0,
# `transformers` will try to import TF as part of its availability checks
# and crash with an unrelated protobuf ImportError.
os.environ.setdefault("USE_TF", "0")

# Belt and suspenders: USE_TF=0 alone isn't reliably respected in this
# environment (transformers still attempts `import tensorflow` internally).
# We never need TF here (PyTorch-only script), so stub it out in sys.modules
# before transformers gets a chance to import the real, broken package.
if "tensorflow" not in sys.modules:
    _fake_tf = types.ModuleType("tensorflow")
    _fake_tf.__version__ = "0.0.0"
    sys.modules["tensorflow"] = _fake_tf

import json
import subprocess

import torch
from PIL import Image

# ---------------------------------------------------------------------
# 0. Config / paths
# ---------------------------------------------------------------------
BASE_MODEL_PATH = "/root/.cache/huggingface/hub/models--Qwen--Qwen2.5-VL-3B-Instruct/snapshots/66285546d2b821cf421d4f5eb2576359d3770cd3"
if not os.path.isdir(BASE_MODEL_PATH):
    print(f"[warn] local snapshot not found at {BASE_MODEL_PATH!r}, "
          f"falling back to HF hub id 'Qwen/Qwen2.5-VL-3B-Instruct'")
    BASE_MODEL_PATH = "Qwen/Qwen2.5-VL-3B-Instruct"

ADAPTER_PATH = "/kaggle/working/trivia_train_run/v33-20260809-194910/checkpoint-30"

DATASET_PATH = "/kaggle/working/TRivia/training/exps/tablevqa_samples.jsonl"
DATASET_DIR = os.path.dirname(DATASET_PATH)

REWARD_PLUGIN_PATH = "/kaggle/working/TRivia/training/exps/trivia_reward_plugin.py"
MS_SWIFT_PATH = "/kaggle/working/TRivia/training/ms-swift"

DEVICE = "cuda:0"
MAX_NEW_TOKENS = 320

PROMPT_TEXT_RAW = (
    "You are an AI specialized in recognizing and extracting tables from images. "
    "Your mission is to analyze the table in the image and reproduce it as a valid "
    "HTML table using <table>, <tr>, <td> (and <th> for header cells) tags. Output "
    "only the HTML table and nothing else.\n<image>"
)
# The trailing "\n<image>" in the training data is ms-swift's inline image
# placeholder. When we build multimodal input for plain `transformers` via
# structured content blocks (type="image" + type="text"), the image is
# already represented by its own content block, so we strip that literal
# placeholder out of the text to avoid double-representing the image.
PROMPT_TEXT = PROMPT_TEXT_RAW.replace("\n<image>", "").strip()

os.environ.setdefault("QA_REWARD_MODEL", "gpt-4o-mini")
if "OPENAI_API_KEY" not in os.environ or not os.environ["OPENAI_API_KEY"]:
    raise RuntimeError(
        "OPENAI_API_KEY is not set in the environment. Set it before running this cell."
    )

# Needed so `trivia_reward_plugin.py`'s internal `from swift.plugin import ORM, orms`
# (and its own sys.path manipulation for data_pipeline) can resolve correctly.
if MS_SWIFT_PATH not in sys.path:
    sys.path.insert(0, MS_SWIFT_PATH)

torch.manual_seed(0)

# ---------------------------------------------------------------------
# 1. Make sure qwen_vl_utils is available (official helper for building
#    Qwen2.5-VL multimodal generate() inputs from chat-template messages)
# ---------------------------------------------------------------------
try:
    from qwen_vl_utils import process_vision_info
except ImportError:
    print("[info] qwen_vl_utils not found, installing...")
    subprocess.run(
        [sys.executable, "-m", "pip", "install", "-q", "qwen-vl-utils"], check=True
    )
    from qwen_vl_utils import process_vision_info

from transformers import AutoProcessor, Qwen2_5_VLForConditionalGeneration
from peft import PeftModel

# ---------------------------------------------------------------------
# 2. Load the first example from the dataset
# ---------------------------------------------------------------------
with open(DATASET_PATH, "r", encoding="utf-8") as f:
    first_line = f.readline()
example = json.loads(first_line)

qa_pairs = example["qa_pairs"]
image_rel_path = example["images"][0]
image_path = (
    image_rel_path
    if os.path.isabs(image_rel_path)
    else os.path.join(DATASET_DIR, image_rel_path)
)

if not os.path.isfile(image_path):
    raise FileNotFoundError(f"Image not found at resolved path: {image_path}")

pil_image = Image.open(image_path).convert("RGB")

print("=" * 80)
print("EXAMPLE")
print("=" * 80)
print(f"Image path : {image_path}")
for i, qa in enumerate(qa_pairs):
    print(f"  Q{i + 1}: {qa['question']}")
    print(f"  A{i + 1}: {qa['answer']}")
print()

# ---------------------------------------------------------------------
# 3. Build the chat-template input (image content block + text block)
# ---------------------------------------------------------------------
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": image_path},
            {"type": "text", "text": PROMPT_TEXT},
        ],
    }
]

# ---------------------------------------------------------------------
# 4. Load processor + base model (float16, sdpa, GPU 0)
# ---------------------------------------------------------------------
print("Loading processor...")
processor = AutoProcessor.from_pretrained(BASE_MODEL_PATH, trust_remote_code=True)

print("Loading base model (float16, sdpa)...")
model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    BASE_MODEL_PATH,
    torch_dtype=torch.float16,
    attn_implementation="sdpa",
    trust_remote_code=True,
    low_cpu_mem_usage=True,
).to(DEVICE)
model.eval()

# Build generate() inputs once — reused for both base and trained passes
text_prompt = processor.apply_chat_template(
    messages, tokenize=False, add_generation_prompt=True
)
image_inputs, video_inputs = process_vision_info(messages)
inputs = processor(
    text=[text_prompt],
    images=image_inputs,
    videos=video_inputs,
    padding=True,
    return_tensors="pt",
).to(DEVICE)


def generate_completion(gen_model, gen_inputs, gen_processor):
    with torch.no_grad():
        generated_ids = gen_model.generate(
            **gen_inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            num_beams=1,
        )
    trimmed_ids = [
        out_ids[len(in_ids):]
        for in_ids, out_ids in zip(gen_inputs.input_ids, generated_ids)
    ]
    decoded = gen_processor.batch_decode(
        trimmed_ids, skip_special_tokens=True, clean_up_tokenization_spaces=False
    )
    return decoded[0].strip()


# ---------------------------------------------------------------------
# 5. BASE model generation (before loading the LoRA adapter)
# ---------------------------------------------------------------------
print("\nGenerating with BASE model...")
base_completion = generate_completion(model, inputs, processor)
torch.cuda.empty_cache()

# ---------------------------------------------------------------------
# 6. Attach the trained LoRA adapter in-memory on the SAME base model
# ---------------------------------------------------------------------
if not os.path.isdir(ADAPTER_PATH):
    raise FileNotFoundError(f"LoRA adapter checkpoint not found at: {ADAPTER_PATH}")

print(f"\nLoading LoRA adapter from: {ADAPTER_PATH}")
peft_model = PeftModel.from_pretrained(model, ADAPTER_PATH)
peft_model.eval()

print("Generating with TRAINED (LoRA-adapted) model...")
trained_completion = generate_completion(peft_model, inputs, processor)
torch.cuda.empty_cache()

# ---------------------------------------------------------------------
# 7. Score both completions with the GRPO reward function
# ---------------------------------------------------------------------
print("\nLoading reward plugin (OpenAIQAF1Score)...")
import importlib.util

spec = importlib.util.spec_from_file_location("trivia_reward_plugin", REWARD_PLUGIN_PATH)
trivia_reward_plugin = importlib.util.module_from_spec(spec)
spec.loader.exec_module(trivia_reward_plugin)
OpenAIQAF1Score = trivia_reward_plugin.OpenAIQAF1Score

scorer = OpenAIQAF1Score()

print("Scoring BASE completion...")
base_reward = scorer(completions=[base_completion], qa_pairs=[qa_pairs])[0]

print("Scoring TRAINED completion...")
trained_reward = scorer(completions=[trained_completion], qa_pairs=[qa_pairs])[0]

# ---------------------------------------------------------------------
# 8. Report
# ---------------------------------------------------------------------
print("\n" + "=" * 80)
print("RESULTS")
print("=" * 80)
print(f"Image path      : {image_path}")
print()
for i, qa in enumerate(qa_pairs):
    print(f"Question {i + 1}    : {qa['question']}")
    print(f"Ground truth {i + 1}: {qa['answer']}")
print()
print("-" * 80)
print("BASE MODEL COMPLETION:")
print("-" * 80)
print(base_completion)
print()
print("-" * 80)
print("TRAINED (LoRA) MODEL COMPLETION:")
print("-" * 80)
print(trained_completion)
print()
print("=" * 80)
print("REWARD SCORES (OpenAIQAF1Score, model={})".format(os.environ["QA_REWARD_MODEL"]))
print("=" * 80)
print(f"  Base model reward    : {base_reward:.4f}")
print(f"  Trained model reward : {trained_reward:.4f}")
print(f"  Delta (trained-base) : {trained_reward - base_reward:+.4f}")
````

````python
path = "/kaggle/working/TRivia/training/exps/trivia_reward_plugin.py"
with open(path) as f:
    src = f.read()

addition = '''

class TableStructureReward(ORM):
    """Self-supervised structural validity reward. No ground truth required.
    Gives partial credit for well-formed table HTML so the model has *some*
    gradient signal even when QA_F1 scores 0, which is the common case with
    a small, sparse-reward dataset.
    """
    def __call__(self, completions, **kwargs):
        return [self._score(c) for c in completions]

    def _score(self, completion: str) -> float:
        try:
            text = completion.strip()
            text = re.sub(r"^```(?:html)?\\s*", "", text)
            text = re.sub(r"```\\s*$", "", text).strip()
            lower = text.lower()

            score = 0.0
            if "<table" not in lower:
                return 0.0
            score += 0.2
            if "</table>" not in lower:
                return score
            score += 0.2

            row_tags = re.findall(r"<tr[\\s>]", lower)
            if len(row_tags) < 2:
                return score
            score += 0.2

            rows = re.split(r"<tr[^>]*>", text, flags=re.IGNORECASE)[1:]
            cell_counts = []
            for row in rows:
                row = row.split("</tr>")[0]
                cells = re.findall(r"<t[dh][\\s>]", row, flags=re.IGNORECASE)
                if cells:
                    cell_counts.append(len(cells))
            if not cell_counts:
                return score
            score += 0.2

            most_common_len, freq = Counter(cell_counts).most_common(1)[0]
            score += 0.2 * (freq / len(cell_counts))

            junk_markers = ["<script", "<style", "<button", "click me",
                            "i\\'m sorry", "as an ai", "this code snippet"]
            if any(m in lower for m in junk_markers):
                score *= 0.5

            return min(score, 1.0)
        except Exception:
            return 0.0


orms["TableStructure"] = TableStructureReward
'''

assert 'orms["TEDS"] = TEDSRewardFunction' in src
src = src.replace('orms["TEDS"] = TEDSRewardFunction', 'orms["TEDS"] = TEDSRewardFunction' + addition)

with open(path, "w") as f:
    f.write(src)
print("added TableStructureReward")
````

````python
import os, sys, json

os.environ.setdefault("USE_TF", "0")
os.environ.setdefault("QA_REWARD_MODEL", "gpt-4o-mini")
os.environ.setdefault("MAX_PIXELS", str(512 * 28 * 28))
os.environ.setdefault("MIN_PIXELS", str(256 * 28 * 28))

MS_SWIFT_PATH = "/kaggle/working/TRivia/training/ms-swift"
if MS_SWIFT_PATH not in sys.path:
    sys.path.insert(0, MS_SWIFT_PATH)

import torch
from swift.llm import PtEngine, InferRequest, RequestConfig

BASE_MODEL = "Qwen/Qwen2.5-VL-3B-Instruct"
ADAPTER_PATH = "/kaggle/working/trivia_train_run_v3/v0-20260810-005802/checkpoint-80"
DATASET_PATH = "/kaggle/working/TRivia/training/exps/tablevqa_samples.jsonl"
DATASET_DIR = os.path.dirname(DATASET_PATH)
REWARD_PLUGIN_PATH = "/kaggle/working/TRivia/training/exps/trivia_reward_plugin.py"

N_SAMPLES = 2
MAX_NEW_TOKENS = 576

PROMPT_TEXT = (
    "You are an AI specialized in recognizing and extracting tables from images. "
    "Your mission is to analyze the table in the image and reproduce it as a valid "
    "HTML table using <table>, <tr>, <td> (and <th> for header cells) tags. "
    "Output only the HTML table and nothing else.\n<image>"
)

rows = []
with open(DATASET_PATH) as f:
    for line in f:
        if line.strip():
            rows.append(json.loads(line))
print(f"{len(rows)} dataset rows")

engine = PtEngine(BASE_MODEL, torch.float16, use_hf=True, attn_impl="sdpa", max_batch_size=0)

import importlib.util
spec = importlib.util.spec_from_file_location("trivia_reward_plugin", REWARD_PLUGIN_PATH)
trivia_reward_plugin = importlib.util.module_from_spec(spec)
spec.loader.exec_module(trivia_reward_plugin)
qa_scorer = trivia_reward_plugin.OpenAIQAF1Score()
teds_scorer = trivia_reward_plugin.TEDSRewardFunction()
struct_scorer = trivia_reward_plugin.TableStructureReward()

request_config = RequestConfig(
    max_tokens=MAX_NEW_TOKENS, temperature=1.0, top_k=50,
    repetition_penalty=1.15, n=N_SAMPLES,
)

def run_pass(label):
    qa_all, teds_all, struct_all = [], [], []
    for i, row in enumerate(rows):
        image_path = os.path.join(DATASET_DIR, row["images"][0])
        qa_pairs = row["qa_pairs"]
        solution = row["solution"]
        req = InferRequest(messages=[{"role": "user", "content": PROMPT_TEXT}], images=[image_path])
        resp = engine.infer([req], request_config)
        completions = [c.message.content.strip() for c in resp[0].choices]

        qa_r = qa_scorer(completions=completions, qa_pairs=[qa_pairs] * len(completions))
        teds_r = teds_scorer(completions=completions, solution=[solution] * len(completions))
        struct_r = struct_scorer(completions=completions)

        qa_all.extend(qa_r)
        teds_all.extend(teds_r)
        struct_all.extend(struct_r)

        print(f"[{label}] row {i}: QA={sum(qa_r)/len(qa_r):.2f}  "
              f"TEDS={sum(teds_r)/len(teds_r):.2f}  struct={sum(struct_r)/len(struct_r):.2f}")
        torch.cuda.empty_cache()
    return qa_all, teds_all, struct_all

print("\n=== BASE PASS ===")
base_qa, base_teds, base_struct = run_pass("BASE")

print(f"\nAttaching adapter: {ADAPTER_PATH}")
engine._add_adapter(ADAPTER_PATH)

print("\n=== TRAINED PASS ===")
trained_qa, trained_teds, trained_struct = run_pass("TRAINED")

def mean(x):
    return sum(x) / len(x)

print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)
print(f"QA_F1     : base={mean(base_qa):.4f}  trained={mean(trained_qa):.4f}  delta={mean(trained_qa)-mean(base_qa):+.4f}")
print(f"TEDS      : base={mean(base_teds):.4f}  trained={mean(trained_teds):.4f}  delta={mean(trained_teds)-mean(base_teds):+.4f}")
print(f"Structure : base={mean(base_struct):.4f}  trained={mean(trained_struct):.4f}  delta={mean(trained_struct)-mean(base_struct):+.4f}")
````
